```python
# Refined strategy to fill missing 'salary' values
# 1. Try to fill using median of 'department' and 'experience'
median_salary_by_dept_exp = df.groupby(['department', 'experience'])['salary'].transform('median')
df['salary'] = df['salary'].fillna(median_salary_by_dept_exp)

# 2. If any NaNs remain, fill using median of 'department' only
median_salary_by_dept = df.groupby('department')['salary'].transform('median')
df['salary'] = df['salary'].fillna(median_salary_by_dept)

# 3. As a final fallback, fill with the overall median salary
df['salary'] = df['salary'].fillna(df['salary'].median())

print("Missing values for salary after refined imputation:")
display(df.isna().sum())
```

```python
## Startergy to fill in rating values..!
# Department and Experiment wise rating filling
median_rating_by_dpt_exp = df.groupby(['department', 'experience'])['rating'].transform("median")
df['rating'] = df['rating'].fillna(median_rating_by_dpt_exp)

#Department wise Rating filling
median_rating_by_dpt = df.groupby('department')['rating'].transform('median')
df['rating'] = df['rating'].fillna(median_rating_by_dpt)

#Rating median !
df['rating'] = df['rating'].fillna(df['rating'].median())
```

```python
## Bonus: projects and experience plus rating.

## Project + Rating and Department wise median
median_bonus_by_rating_project_dept = df.groupby(['department','projects_completed', 'rating'])['bonus'].transform('median')
df['bonus'] = df['bonus'].fillna(median_bonus_by_rating_project_dept)

## Project wise median
median_bonus_by_projects_rating = df.groupby(['projects_completed','rating'])['bonus'].transform('median')
df['bonus'] = df['bonus'].fillna(median_bonus_by_projects_rating)
## Bonus wise Median

df['bonus']  = df['bonus'].fillna(df['bonus'].median())
```

```python
# Define conditions for salary categories
conditions = [
    df['salary'] > 1000000,
    (df['salary'] > 600000) & (df['salary'] <= 1000000)
]

# Define corresponding choices
choices = ['High', 'Medium']

# Use np.select to assign salary categories, with 'Low' as default
df['salary_category'] = np.select(conditions, choices, default='Low')

# Display the counts of each category to verify
display(df)
```

```python
# 17.Create a column called performance.Rules: Rating ≥ 4.7 and Projects ≥ 10 → Excellent
# Rating ≥ 4.3 → Good
# Otherwise → Average

conditions = [
    ((df['rating'] >= 4.7) & (df['projects_completed'] >= 10)),
    (df['rating'] >= 4.3)
]

choices = ['Excellent', 'Good']

df['performace'] = np.select(conditions, choices, default='Average')

display(df[['name', 'rating', 'projects_completed', 'performace']].head())
```

```python
# 18.Create a column called experience_level.
# Rules:
# 0–2 → Beginner
# 3–5 → Junior
# 6–10 → Mid
# 11+ → Senior

conditions = [
    ((df['experience'] <=2) & (df['experience'] >=0)),
    ((df['experience'] <=5) & (df['experience'] >=3)),
    ((df['experience'] <=10) & (df['experience'] >=6))
]

choices = ['Beginner', 'Junior', 'Mid']

df['experience_level'] = np.select(conditions, choices, default='Senior')

display(df[['name', 'age', 'experience', 'experience_level']].head())


display(df['experience_level'].value_counts())
```
